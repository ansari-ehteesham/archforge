# Chapter 01: LangChain Basics

```python
from langchain.agents import create_agent
import json
```

```python
# Define a basic tool

def check_weather(city_name:str) -> str:
    """
    Get the wheather information of a city
    
    Args:
        city_name: str - Give the City Name
    """
    
    return f"It is Rainy today in {city_name}"
```

Model Name Format is

**provider-name:model-name**

```python
# Define the Agent with Model Name, Tools List and System Prompt

agent = create_agent(
    model="google_genai:gemini-3.5-flash-lite",
    tools=[check_weather],
    system_prompt="You are a helpful assistant"
)
```

### User Prompt Message Structure

{
    "messages": [{
        "role": "user",
        "content": "User Prompt"
    }]
}

```python
# Now, we'll invoke the model and give it a user prompt

result = agent.invoke(
    input={
        "messages": [{
            "role": "user",
            "content": "What is the weather of Delhi, India?"
        }]
    }
)
```
